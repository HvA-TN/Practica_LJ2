```python
import numpy as np
import matplotlib.pyplot as plt

# Publicatie-achtige opmaak
plt.rcParams.update({
    "font.size": 11,
    "axes.labelsize": 12,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "legend.fontsize": 10,
    "axes.linewidth": 1.0,
})

# Meetdata (x, y) met onzekerheden
xdata = np.array([1, 2, 3, 4, 5])
ydata = np.array([1.1, 1.9, 3.2, 3.9, 5.1])
yerr  = np.array([0.1] * 5)

# Figuur maken
fig, ax = plt.subplots(figsize=(6, 4))

ax.errorbar(
    xdata,
    ydata,
    yerr=yerr,
    fmt="o",
    capsize=3,
    label="Meetdata",
    color= 'black',
)

ax.set_xlabel("Afstand (meter)")
ax.set_ylabel("Tijd (seconde)")

ax.tick_params(direction="in", top=True, right=True)
ax.minorticks_on()

fig.tight_layout()
plt.show()
```
